```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import matplotx
import pandas as pd
from pathlib import Path

plt.style.use(matplotx.styles.ayu['light'])
DATA_DIR = Path("data")
```

```python
Y = pd.read_csv(DATA_DIR / "small_movies_Y.csv", header=None).to_numpy()
R = pd.read_csv(DATA_DIR / "small_movies_R.csv", header=None).to_numpy()
movie_df = pd.read_csv(DATA_DIR / "small_movie_list.csv", index_col=0).to_numpy()
X = pd.read_csv(DATA_DIR / "small_movies_X.csv", header=None).to_numpy()
W = pd.read_csv(DATA_DIR / "small_movies_W.csv", header=None).to_numpy()
b = pd.read_csv(DATA_DIR / "small_movies_b.csv", header=None).to_numpy()

print("ratings_df:", Y.shape)
print("rated_df:", R.shape)
print("movie_df:", movie_df.shape)
print("movie_features_df:", X.shape)
print("user_features_df:", W.shape)
print("user_bias_df:", b.shape)
```

```python
num_movies, num_features = X.shape
num_users, num_features = W.shape

print(num_movies, num_users, num_features)
X
```

```python
import re
import numpy as np

# Find row indices where user 0 has provided a rating
rated_indices = np.where(R[:, 0] == 1)[0]

# Extract titles for those specific movies 
# (Assuming the title is the 3rd column in movie_df based on your dataset shape)
rated_titles = movie_df[rated_indices, 2]

# Parse the 4-digit year enclosed in parentheses from the title string
def get_year(title):
    match = re.search(r'\((\d{4})\)', str(title))
    return int(match.group(1)) if match else float('inf')

# Locate and print the movie with the minimum year
if len(rated_titles) > 0:
    earliest = min(rated_titles, key=get_year)
    print(f"Earliest movie rated by user 0: {earliest}")
else:
    print("User 0 has not rated any movies.")
```

```python
def test_plot(movieidx, useridx):
    # Support both NumPy arrays and TensorFlow variables
    m_vec = X[movieidx].numpy() if hasattr(X, 'numpy') else X[movieidx]
    u_vec = W[useridx].numpy() if hasattr(W, 'numpy') else W[useridx]
    b_val = b[0, useridx].numpy() if hasattr(b, 'numpy') else b[0, useridx]
    
    # 2D slices for plotting
    movie_2d = m_vec[:2]
    user_2d = u_vec[:2]
    
    is_rated = bool(R[movieidx, useridx])
    actual = Y[movieidx, useridx]
    
    # Calculate prediction using ALL 10 features, not just the 2D slice
    pred = np.dot(m_vec, u_vec) + b_val
    error = np.abs(pred - actual) if is_rated else None

    print(f"User rated? {is_rated}")
    print("Movie features (2D): ", movie_2d)
    print("User features (2D): ", user_2d)
    print(f"Bias: {b_val:.4f}")
    print(f"Predicted Rating: {pred:.4f}")
    if is_rated:
        print(f"Actual rating: {actual}")

    # Plotting
    fig, ax1 = plt.subplots(1, 1, figsize=(6, 5))
    
    # Quiver plots using high-contrast colors
    ax1.quiver(0, 0, movie_2d[0], movie_2d[1], angles="xy", scale_units="xy", 
               color='deepskyblue', scale=1, label=f"Movie {movieidx}")
    ax1.quiver(0, 0, user_2d[0], user_2d[1], angles="xy", scale_units="xy", 
               color='orange', scale=1, label=f"User {useridx}")
    
    # Dynamic axis limits based on vector magnitude
    max_val = max(abs(movie_2d[0]), abs(movie_2d[1]), abs(user_2d[0]), abs(user_2d[1])) * 1.5
    ax1.set_xlim(-max_val, max_val)
    ax1.set_ylim(-max_val, max_val)
    
    # Origin lines to clearly separate quadrants
    ax1.axhline(0, color='gray', linestyle='--', alpha=0.5, linewidth = 1.2)
    ax1.axvline(0, color='gray', linestyle='--', alpha=0.5, linewidth = 1.2)
    
    # Title acting as the error label
    title = f"Movie {movieidx} vs User {useridx} (First 2 Latent Features)"
    if is_rated:
        title += f"\nActual: {actual:.1f} | Pred: {pred:.2f} | Error: {error:.2f}"
    else:
        title += f"\n(Unrated) | Pred: {pred:.2f}"
    ax1.set_title(title, pad=12)

    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='upper right')
    plt.show()

test_plot(4, 0)
```

X = movie features \
no. of movie features = 10
$$
\begin{align}
\hat{Y} &= X \times W^T + b \\
L_2 &= \frac{\lambda}{2} \left( \sum W^2 + \sum X^2 \right) \\
J &= \frac{1}{2} \sum \left(\hat{Y}^2 - Y\right) + L_2
\end{align}
$$

```python
X.shape, W.shape, Y.shape, b.shape
# X * W^T + b

def J(l):
    Y_cap = X @ tf.transpose(W) + b

    l2 = .5 * l * (tf.reduce_sum(tf.square(W)) + tf.reduce_sum(tf.square(X)))
    err = .5 * tf.reduce_sum(tf.square(Y_cap - Y) * R) + l2
    return err

J(0)
```

```python
tf.random.set_seed(1234)
W = tf.Variable(tf.random.normal((num_users, num_features), dtype = tf.float64), name = "W")
X = tf.Variable(tf.random.normal((num_movies, num_features), dtype = tf.float64), name = "X")
b = tf.Variable(tf.random.normal((1, num_users), dtype = tf.float64), name = "b")

optimizer = tf.keras.optimizers.Adam(learning_rate= 1e-1)
```

```python
# List to store the cost values for plotting
J_history = []

# Training
itrs = 2000
lambda_ = 1

for itr in range(itrs + 1):
    with tf.GradientTape() as tape:
        cost_value = J(lambda_)
        
    grads = tape.gradient(cost_value, [X, W, b])
    optimizer.apply_gradients(zip(grads, [X, W, b]))

    # Append the numpy value of the cost to our history list
    J_history.append(cost_value.numpy())

    # Print every 200 iterations to avoid output spam
    if itr % 200 == 0:
        print(f"Training loss at iteration {itr}: {cost_value.numpy():0.1f}")
```

```python
J_history = np.array(J_history)

# Plotting the Learning Curve
fig, ax = plt.subplots(1, 1, figsize=(8, 5))

# Using 'deepskyblue' to match your high-contrast theme preference
ax.plot(J_history, color='deepskyblue', linewidth=2, label="Training Loss")

ax.set_title("Cost Function vs. Iterations", pad=15, fontsize=14)
ax.set_xlabel("Iteration", fontsize=12)
ax.set_ylabel("Cost (J)", fontsize=12)

# Faint grid for readability without overwhelming the dark theme
ax.grid(True, alpha=0.3, linestyle='--')
ax.legend()
```
