```python
import tensorflow as tf
import numpy as np
```

```python
tf.random.set_seed(1234)
a = tf.Variable(tf.random.normal((1, 1), dtype = tf.float64), name = "a")
a
```

```python
with tf.GradientTape() as tape:
    cost = a ** 2

grad = tape.gradient(cost, [a])
print(grad)
```
